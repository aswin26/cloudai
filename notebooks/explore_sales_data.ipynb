Sales Data Exploration

This notebook explores sales_data.csv interactively using Pandas
We use the same dataset that data_tool.py processess, but here we explore it visually and ask ad-hoc question

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# display charts inline
%matplotlib inline

sns.set_style("whitegrid")

print("Libraries loaded successfully")

```

```python
df = pd.read_csv("../sample_data/sales_data.csv")
```

```python
print(f"Loaded {len(df)} rows and {len(df.columns)} columns")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df)
```

```python
revenue_by_product = (
    df.groupby("product")["unit_price"]
    .sum()
    .sort_values(ascending=False)
)
print(revenue_by_product)
```

```python
fig, ax = plt.subplots(figsize=(10,6))
revenue_by_product.plot(kind = "bar", ax=ax, color = "steelblue", edgecolor="black")

ax.set_title("Total revenue by product", fontsize = 14, fontweight= "bold")
ax.set_xlabel("Product",fontsize = 12)
ax.set_ylabel("Revenue($)",fontsize=12)
ax.tick_params(axis="x",rotation = 45)

plt.tight_layout()
plt.show()
```

```python
revenue_by_region = (
    df.groupby("region")["unit_price"]
    .sum()
    .sort_values(ascending=False)
)
# print(revenue_by_product)
fig, ax = plt.subplots(figsize=(8,5))
revenue_by_region.plot(kind = "barh", ax=ax, color = "coral", edgecolor="black")

ax.set_title("Total revenue by Region", fontsize = 14, fontweight= "bold")
ax.set_xlabel("Region",fontsize = 12)
ax.set_ylabel("Revenue($)",fontsize=12)
# ax.tick_params(axis="x",rotation = 45)

plt.tight_layout()
plt.show()
```

# Distribution Analysis

```python
fig, ax = plt.subplots(figsize=(10,6))
ax.hist(df["unit_price"],bins = 20, color = "mediumseagreen",edgecolor="black",alpha=0.7)
ax.set_title("Revenue Distribution", fontsize = 14, fontweight= "bold")
ax.set_xlabel("Revenue",fontsize = 12)
ax.set_ylabel("Frequency($)",fontsize=12)

plt.tight_layout()
plt.show()
```
